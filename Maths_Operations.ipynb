```python
# Basic Mathematical Operations
```

```python
num1 = float(input("Enter first number: "))
operator = input("Enter Mathematical Operator (+, -, *, or /): ")
num2 = float(input("Enter second number: "))

match operator:
  case  "+":
    solution = num1 + num2
  case  "-":
    solution = num1 - num2
  case "*":
    solution = num1 * num2
  case  "/":
    solution = num1/num2
  case _:
    solution = "invalid operator"

print(solution)
```
